```python
import matplotlib.pyplot as plt
import numpy as np

# Dati estratti dal tuo LaTeX (Table 1: Pure spatial interpolation)
h = np.array([1.2500e-01, 6.2500e-02, 3.1250e-02, 1.5625e-02])
error_L2 = np.array([1.3329e-02, 3.3602e-03, 8.4181e-04, 2.1056e-04])
error_H1 = np.array([2.5308e-01, 1.2607e-01, 6.2977e-02, 3.1481e-02])

plt.figure(figsize=(7, 6))

# Grafico dell'errore L2 e H1
plt.loglog(h, error_L2, '-o', label=r'$L^2$ Error (Slope $\approx$ 2.0)', color='#005088', linewidth=2)
plt.loglog(h, error_H1, '-s', label=r'$H^1$ Error (Slope $\approx$ 1.0)', color='#11caa0', linewidth=2)

# Linee di riferimento teoriche per l'occhio del revisore
plt.loglog(h, 0.8 * (h**2), '--', color='gray', label=r'$\mathcal{O}(h^2)$ Reference')
plt.loglog(h, 2.0 * h, ':', color='gray', label=r'$\mathcal{O}(h)$ Reference')

# Formattazione assi e label (Rigorosamente in inglese per le slide)
plt.xlabel('Mesh Size ($h$)', fontsize=12)
plt.ylabel('Discretization Error', fontsize=12)
plt.title('Spatial Convergence Analysis (MMS)', fontsize=14, fontweight='bold', pad=15)
plt.grid(True, which="both", ls="--", alpha=0.5)
plt.legend(fontsize=11, loc='upper left')

# Salva l'immagine per le slide
plt.tight_layout()
plt.savefig('mms_convergence.png', dpi=300)
plt.show()
```
